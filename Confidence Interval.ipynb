
# Confidence Interval Estimation

This notebook demonstrates how to calculate confidence intervals for a sample mean using Python. Both cases—using the sample standard deviation (t-distribution) and using the known population standard deviation (normal distribution)—are covered.

The example deals with durability data (in millions of characters) for print-heads, a small-sample scenario where t-distribution is appropriate.


```

# Sample durability data (print-heads)
durability_data = [1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32,
                   0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29]
```


We have a small dataset of 15 samples representing the durability of print-heads. This small sample size and unknown population variance typically require the use of the t-distribution for confidence interval estimation.


```

import numpy as np

# Calculate sample mean and standard deviation
sample_mean = np.mean(durability_data)
sample_std = np.std(durability_data, ddof=1)  # sample std dev
sample_size = len(durability_data)

print(f'Sample Mean: {sample_mean}')
print(f'Sample Standard Deviation: {sample_std}')
```


The sample mean is our best estimate of the population mean. The sample standard deviation estimates the spread, using degree of freedom correction (ddof=1).


```

from scipy.stats import t

# Confidence level
confidence = 0.99

# Degrees of freedom
df = sample_size - 1

# t critical value for 99% confidence
t_critical = t.ppf((1 + confidence) / 2, df)

# Margin of error
margin_error = t_critical * (sample_std / np.sqrt(sample_size))

# Confidence interval
ci_lower = sample_mean - margin_error
ci_upper = sample_mean + margin_error

print(f'99% confidence interval using sample std dev: ({ci_lower:.4f}, {ci_upper:.4f})')
```


The 99% confidence interval calculated with the t-distribution accounts for the sample size and variability, providing a range likely to contain the true population mean.


```

from scipy.stats import norm

# Given population standard deviation
pop_std = 0.2

# Z critical value for 99% confidence
z_critical = norm.ppf((1 + confidence) / 2)

# Margin of error with population std
margin_error_pop = z_critical * (pop_std / np.sqrt(sample_size))

# Confidence interval
ci_lower_pop = sample_mean - margin_error_pop
ci_upper_pop = sample_mean + margin_error_pop

print(f'99% confidence interval using population std dev: ({ci_lower_pop:.4f}, {ci_upper_pop:.4f})')
```


When the population standard deviation is known, the normal distribution (Z-distribution) is used to calculate the confidence interval, usually resulting in a narrower interval.
